```python

import torch
import diffusers

print("PyTorch:", torch.__version__)
print("Diffusers:", diffusers.__version__)
print("GPU:", torch.cuda.get_device_name(0))
print("CUDA available:", torch.cuda.is_available())
```

```python
import torch
from diffusers import WanPipeline
from diffusers.utils import export_to_video

model_id = "Wan-AI/Wan2.1-T2V-1.3B-Diffusers"

pipe = WanPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
)

pipe.to("cuda")
print("Wan model loaded successfully!")
```

```python

prompt = """
Beautiful original Japanese anime fantasy animation.
Golu, a cute little blue baby elephant with big brown eyes
and a yellow scarf, walks through a magical moonlit forest.
He blinks, moves his ears, turns toward a glowing ancient tree,
and takes two small steps forward. Fireflies drift around him,
leaves sway gently, and magical light sparkles.
Smooth continuous character animation, expressive face,
consistent character design, family-friendly, cinematic lighting.
No text, no subtitles.
"""

with torch.inference_mode():
    result = pipe(
        prompt=prompt,
        negative_prompt="still image, frozen character, blurry, text, watermark",
        height=480,
        width=832,
        num_frames=33,
        num_inference_steps=20,
        guidance_scale=5.0,
    )

frames = result.frames[0]
export_to_video(frames, "golu_test.mp4", fps=16)

print("Video saved: golu_test.mp4")
```
