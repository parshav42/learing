```python
import pandas as pd

data = {
    "date": pd.date_range("2026-01-01", periods=6, freq="D"),
    "sales": [100, 200, 150, 300, 250, 400]
}

df = pd.DataFrame(data)
df = df.set_index("date")

print(df)
```
