```python
x = torch.tensor([1., 2., 3., 4., 5., 6., 7., 8.])
y = torch.tensor([5., 9., 13., 17., 21., 25., 29., 33.])
```

```python
x = torch.tensor([1., 2., 3., 4., 5., 6., 7., 8.], requires_grad = True)
y = torch.tensor([5., 9., 13., 17., 21., 25., 29., 33.],requires_grad = True)
w = torch.tensor(1.0,requires_grad = True)
b = torch.tensor(1.0,requires_grad = True)
lr =0.04
for i in range(200):
    pre = w * x + b

    loss = (pre - y)**2

    loss =loss.mean()

    loss.backward()

    with torch.no_grad():
        # w = w - lr * w.grad()
        w -=  lr * w.grad
        b -=  lr * b.grad
    w.grad.zero_()
    b.grad.zero_()


```
