```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn import datasets

iris= datasets.load_iris()

# print(iris.data)
print(iris.target)
```

```python
print(iris.data.shape)
print(iris.target.shape)
```

```python
x = np.array([1,2,3,4])

print(x.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(iris.data ,iris.target ,test_size = 0.4, random_state = True)
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()  # This is an estimator
model.fit(x_train, y_train)
predictions = model.predict(x_test)
```
